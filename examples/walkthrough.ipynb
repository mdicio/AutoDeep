# AutoDeep walkthrough

Benchmark gradient-boosted trees and deep tabular models on a few datasets, with a runtime estimate before training and a leaderboard after.

Everything here runs offline on CPU in a couple of minutes (built-in datasets, small search budget).

````python
import pandas as pd

from autodeep import AutoRunner, ResultsAnalyzer, list_models

pd.set_option("display.width", 200)
pd.set_option("display.max_columns", 50)
````

## Available models

Names, families and presets (`fast`, `default`, `deep`, `all`).

````python
pd.DataFrame(list_models())[["name", "display_name", "family", "speed"]]
````

## Define the experiment

Only `path` and `target` are required for your own files; `builtin:` datasets need neither a target nor a download.

```python
"my_data": {"path": "data/my_data.csv", "target": "label", "drop_columns": ["id"]}
```

````python
runner = AutoRunner(
    datasets={
        "breast_cancer": {"path": "builtin:breast_cancer"},  # binary
        "iris": {"path": "builtin:iris"},  # multiclass
        "diabetes": {"path": "builtin:diabetes"},  # regression
    },
    models=["xgb", "catboost", "mlp", "categoryembedding"],  # or a preset: "fast", "default", "deep", "all"
    max_evals=5,
    max_epochs=30,
    output_folder="autodeep_results",
)
````

## Estimate the runtime first

````python
runner.estimate()
````

## Run

````python
leaderboard = runner.run()
leaderboard
````

## Inspect the saved results

`ResultsAnalyzer` reads the output folder, so this also works from a later session.

````python
analyzer = ResultsAnalyzer(runner.results_path)
analyzer.view_performance(dataset_name="breast_cancer")
````

````python
analyzer.view_parameters(model_name="xgb")
````
